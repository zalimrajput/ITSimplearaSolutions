```python

# ================================================================
# AGRIPAK FARMS - LOW-YIELD RISK PREDICTION
# MACHINE LEARNING INTERNSHIP - TASK 01 / WEEK 1
#
# Dataset: crop_data.csv
# Target: low_yield_risk
# Model: Logistic Regression
# Split: 80% Training / 20% Testing
# Random State: 42
#
# Leakage-aware development pipeline:
# - Stratified train-test split
# - Training-only correlation analysis
# - Numerical and categorical correlation
# - Train-test overlap check
# - Training-only preprocessing
# - Unseen test-set evaluation
#
# ================================================================


# ================================================================
# 1. INSTALL LIBRARIES
# ================================================================

!pip install -q pandas numpy scikit-learn matplotlib seaborn joblib


# ================================================================
# 2. IMPORT LIBRARIES
# ================================================================

import os
import json
import shutil
import joblib
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

RANDOM_STATE = 42
TARGET_COLUMN = "low_yield_risk"
THRESHOLD = 0.90


# ================================================================
# 3. CREATE PROJECT STRUCTURE
# ================================================================

PROJECT_DIR = "/content/AgriPak-Low-Yield-Risk-Prediction"

folders = [
    "data/raw",
    "data/processed",
    "src",
    "notebooks",
    "models",
    "reports"
]

for folder in folders:
    os.makedirs(
        os.path.join(PROJECT_DIR, folder),
        exist_ok=True
    )

print("Project structure created!")


# Remove previous generated reports to prevent stale files
# from being included in the new project ZIP.

for filename in os.listdir(f"{PROJECT_DIR}/reports"):

    file_path = os.path.join(
        PROJECT_DIR,
        "reports",
        filename
    )

    if os.path.isfile(file_path):
        os.remove(file_path)


# ================================================================
# 4. UPLOAD DATASET
# ================================================================

print("\nUpload your crop_data.csv file")

uploaded = files.upload()

csv_files = [
    name for name in uploaded.keys()
    if name.lower().endswith(".csv")
]

if not csv_files:
    raise ValueError("Please upload a CSV file.")

df = pd.read_csv(csv_files[0])

# Record original dataset size before cleaning.
original_rows = len(df)

# Clean column names.
df.columns = df.columns.astype(str).str.strip()

# Validate target column.
if TARGET_COLUMN not in df.columns:

    raise ValueError(
        f"Target '{TARGET_COLUMN}' not found. "
        f"Available columns: {df.columns.tolist()}"
    )

# Save original uploaded dataset.
df.to_csv(
    f"{PROJECT_DIR}/data/raw/crop_data.csv",
    index=False
)

print("\n========== DATASET INFORMATION ==========")

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Records:")
display(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())


# ================================================================
# 5. VALIDATE TARGET
# ================================================================

cleaning_log = []

# Convert target into numeric format.
df[TARGET_COLUMN] = pd.to_numeric(
    df[TARGET_COLUMN],
    errors="coerce"
)

# Count missing or invalid target values.
missing_target_count = int(
    df[TARGET_COLUMN].isna().sum()
)

# Remove rows with missing target.
df = df.dropna(
    subset=[TARGET_COLUMN]
).copy()

cleaning_log.append(
    f"Removed {missing_target_count} rows with missing or invalid target values."
)

# Ensure target contains only binary values.
if not set(df[TARGET_COLUMN].unique()).issubset({0, 1}):

    raise ValueError(
        "low_yield_risk must contain only 0 and 1."
    )

df[TARGET_COLUMN] = df[TARGET_COLUMN].astype(int)

if df[TARGET_COLUMN].nunique() != 2:

    raise ValueError(
        "Both target classes must be present."
    )


# ================================================================
# 6. DATA CLEANING
# ================================================================

# Remove duplicate records after target validation.
duplicates = int(df.duplicated().sum())

df = df.drop_duplicates().copy()

cleaning_log.append(
    f"Removed {duplicates} duplicate rows."
)

# Convert numeric-looking columns.
for col in df.columns:

    if col == TARGET_COLUMN:
        continue

    if df[col].dtype == "object":

        converted = pd.to_numeric(
            df[col],
            errors="coerce"
        )

        original_non_missing = df[col].notna().sum()

        if original_non_missing > 0:

            ratio = (
                converted.notna().sum()
                / original_non_missing
            )

            if ratio >= 0.90:

                df[col] = converted

                cleaning_log.append(
                    f"Converted {col} to numeric."
                )

# Validate final target.
df[TARGET_COLUMN] = df[TARGET_COLUMN].astype(int)

cleaned_rows = len(df)

removed_rows = original_rows - cleaned_rows

print("\n========== CLEANING RESULTS ==========")

print("Original records:", original_rows)
print("Cleaned records:", cleaned_rows)
print("Removed records:", removed_rows)
print("Removed duplicates:", duplicates)
print("Removed missing/invalid targets:", missing_target_count)

print("\nCleaning decisions:")

for decision in cleaning_log:
    print("-", decision)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

# Save cleaned dataset.
df.to_csv(
    f"{PROJECT_DIR}/data/processed/cleaned_data.csv",
    index=False
)


# ================================================================
# 7. EXPLORATORY DATA ANALYSIS
# ================================================================

print("\n========== TARGET DISTRIBUTION ==========")

print(df[TARGET_COLUMN].value_counts())

print("\nTarget Percentage:")

print(
    df[TARGET_COLUMN].value_counts(normalize=True) * 100
)

plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x=TARGET_COLUMN
)

plt.title("Low Yield Risk Distribution")
plt.xlabel("Low Yield Risk (0 = No, 1 = Yes)")
plt.ylabel("Number of Records")

plt.tight_layout()

plt.savefig(
    f"{PROJECT_DIR}/reports/target_distribution.png"
)

plt.show()


# ================================================================
# 8. SEPARATE FEATURES AND TARGET
# ================================================================

X = df.drop(
    columns=[TARGET_COLUMN]
).copy()

y = df[TARGET_COLUMN].copy()

print("\n========== FEATURES AND TARGET ==========")

print("Features:")
print(X.columns.tolist())

print("\nTarget:", TARGET_COLUMN)

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)


# ================================================================
# 9. STRATIFIED TRAIN-TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("\n========== TRAIN TEST SPLIT ==========")

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))


# ================================================================
# 10. SAVE SPLIT FILES
# ================================================================

processed_dir = f"{PROJECT_DIR}/data/processed"

X_train.to_csv(
    f"{processed_dir}/X_train.csv",
    index=False
)

X_test.to_csv(
    f"{processed_dir}/X_test.csv",
    index=False
)

pd.DataFrame({
    TARGET_COLUMN: y_train
}).to_csv(
    f"{processed_dir}/y_train.csv",
    index=False
)

pd.DataFrame({
    TARGET_COLUMN: y_test
}).to_csv(
    f"{processed_dir}/y_test.csv",
    index=False
)

print("\nAll four split files saved!")


# ================================================================
# 11. TRAIN-TEST OVERLAP CHECK
# ================================================================

print("\n========== TRAIN-TEST OVERLAP CHECK ==========")

# Combine features and target for exact record comparison.
train_records = pd.concat(
    [
        X_train.reset_index(drop=True),
        y_train.reset_index(drop=True)
    ],
    axis=1
)

test_records = pd.concat(
    [
        X_test.reset_index(drop=True),
        y_test.reset_index(drop=True)
    ],
    axis=1
)

train_hashes = set(
    pd.util.hash_pandas_object(
        train_records,
        index=False
    )
)

test_hashes = set(
    pd.util.hash_pandas_object(
        test_records,
        index=False
    )
)

overlap = train_hashes.intersection(test_hashes)

overlap_count = len(overlap)

print("Exact overlapping records:", overlap_count)

if overlap_count == 0:

    print(
        "No exact duplicate records found across train and test."
    )

else:

    print(
        "WARNING: Exact duplicate records found across splits!"
    )


# ================================================================
# 12. TRAINING-ONLY CORRELATION ANALYSIS
#     NUMERICAL + CATEGORICAL FEATURES
# ================================================================

print(
    "\n========== TRAINING-ONLY CORRELATION ANALYSIS =========="
)

# IMPORTANT:
# Use X_train and y_train only.
# Do not use X_test or y_test for feature investigation.
#
# This prevents test-set information from influencing
# feature investigation during model development.

X_analysis = X_train.copy()
y_analysis = y_train.copy()

# Identify numerical features.
numeric_features = X_analysis.select_dtypes(
    include=["number"]
).columns.tolist()

# Identify categorical features.
categorical_features = X_analysis.select_dtypes(
    include=["object", "category", "bool", "string"]
).columns.tolist()

print("\nNumerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)

all_correlation_results = []

categorical_detail_results = []


# ------------------------------------------------
# A. NUMERICAL FEATURE CORRELATION
# ------------------------------------------------

for feature in numeric_features:

    correlation = X_analysis[feature].corr(
        y_analysis
    )

    if pd.notna(correlation):

        all_correlation_results.append({

            "Feature": feature,

            "Feature_Type": "Numerical",

            "Correlation": float(correlation),

            "Absolute_Correlation": float(
                abs(correlation)
            )
        })


# ------------------------------------------------
# B. CATEGORICAL FEATURE CORRELATION
# ------------------------------------------------

for feature in categorical_features:

    encoded = pd.get_dummies(
        X_analysis[feature],
        prefix=feature,
        dtype=int
    )

    category_correlations = []

    for category_column in encoded.columns:

        correlation = encoded[category_column].corr(
            y_analysis
        )

        if pd.isna(correlation):
            continue

        category_correlations.append(
            abs(correlation)
        )

        categorical_detail_results.append({

            "Original_Feature": feature,

            "Category": category_column,

            "Correlation": float(correlation),

            "Absolute_Correlation": float(
                abs(correlation)
            )
        })

    # Feature-level association is represented by
    # its strongest individual category correlation.

    if category_correlations:

        max_correlation = max(
            category_correlations
        )

        all_correlation_results.append({

            "Feature": feature,

            "Feature_Type": "Categorical",

            "Correlation": np.nan,

            "Absolute_Correlation": float(
                max_correlation
            )
        })


# ------------------------------------------------
# C. CREATE FINAL CORRELATION DATAFRAMES
# ------------------------------------------------

all_correlation_df = pd.DataFrame(
    all_correlation_results
)

categorical_detail_df = pd.DataFrame(
    categorical_detail_results
)

if not all_correlation_df.empty:

    all_correlation_df = all_correlation_df.sort_values(
        by="Absolute_Correlation",
        ascending=False
    ).reset_index(drop=True)

if not categorical_detail_df.empty:

    categorical_detail_df = categorical_detail_df.sort_values(
        by="Absolute_Correlation",
        ascending=False
    ).reset_index(drop=True)

print("\n========== FEATURE-LEVEL RESULTS ==========")

display(all_correlation_df)

print("\n========== CATEGORY-LEVEL RESULTS ==========")

display(categorical_detail_df)


# ------------------------------------------------
# D. IDENTIFY POTENTIAL LEAKAGE CANDIDATES
# ------------------------------------------------

flagged_features = all_correlation_df[
    all_correlation_df["Absolute_Correlation"] > THRESHOLD
].copy()

print("\n========== POTENTIAL LEAKAGE FEATURES ==========")

if not flagged_features.empty:

    display(flagged_features)

else:

    print(
        "No training numerical or categorical feature "
        "exceeded the absolute correlation threshold of 0.90."
    )

print(
    "\nNOTE: Correlation screening does not prove "
    "that the dataset is leakage-free."
)


# ------------------------------------------------
# E. SAVE CORRELATION REPORTS
# ------------------------------------------------

all_correlation_df.to_csv(
    f"{PROJECT_DIR}/reports/all_feature_correlation.csv",
    index=False
)

categorical_detail_df.to_csv(
    f"{PROJECT_DIR}/reports/categorical_correlation_details.csv",
    index=False
)

flagged_features.to_csv(
    f"{PROJECT_DIR}/reports/potential_leakage_features.csv",
    index=False
)


# ------------------------------------------------
# F. VISUALIZE FEATURE ASSOCIATION STRENGTH
# ------------------------------------------------

plt.figure(figsize=(10, 6))

sns.barplot(
    data=all_correlation_df,
    x="Absolute_Correlation",
    y="Feature"
)

plt.axvline(
    THRESHOLD,
    color="red",
    linestyle="--",
    label="Investigation threshold (0.90)"
)

plt.title(
    "Training Feature Associations with Low Yield Risk"
)

plt.xlabel(
    "Absolute Pearson / Maximum Category Correlation"
)

plt.ylabel("Feature")

plt.legend()

plt.tight_layout()

plt.savefig(
    f"{PROJECT_DIR}/reports/all_feature_correlation.png"
)

plt.show()

print(
    "\nTraining-only correlation analysis completed!"
)


# ================================================================
# 13. IDENTIFY NUMERICAL AND CATEGORICAL FEATURES
# ================================================================

# Feature types are identified from training data.
# These same columns are used for test preprocessing.

numeric_features = X_train.select_dtypes(
    include=["number"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool", "string"]
).columns.tolist()

print("\nNumerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)


# ================================================================
# 14. BUILD PREPROCESSING PIPELINE
# ================================================================

numeric_transformer = Pipeline(
    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="median"
            )
        ),

        (
            "scaler",
            StandardScaler()
        )

    ]
)

categorical_transformer = Pipeline(
    steps=[

        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            )
        ),

        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore"
            )
        )

    ]
)

preprocessor = ColumnTransformer(
    transformers=[

        (
            "numeric",
            numeric_transformer,
            numeric_features
        ),

        (
            "categorical",
            categorical_transformer,
            categorical_features
        )

    ]
)


# ================================================================
# 15. LOGISTIC REGRESSION BASELINE MODEL
# ================================================================

baseline_model = Pipeline(
    steps=[

        (
            "preprocessor",
            preprocessor
        ),

        (
            "classifier",
            LogisticRegression(
                max_iter=2000,
                random_state=RANDOM_STATE
            )
        )

    ]
)

print("\n========== MODEL TRAINING ==========")

# Fit preprocessing and model on training data only.
baseline_model.fit(
    X_train,
    y_train
)

print(
    "Logistic Regression trained successfully!"
)


# ================================================================
# 16. MODEL PREDICTION
# ================================================================

# Transform test data using fitted training preprocessing.
# No fitting takes place on the test set.

y_pred = baseline_model.predict(
    X_test
)


# ================================================================
# 17. MODEL EVALUATION
# ================================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    pos_label=1,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    pos_label=1,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    pos_label=1,
    zero_division=0
)

print("\n========== BASELINE MODEL RESULTS ==========")

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        labels=[0, 1],
        target_names=[
            "No Low Yield Risk",
            "Low Yield Risk"
        ],
        zero_division=0
    )
)


# ------------------------------------------------
# SAVE BASELINE METRICS
# ------------------------------------------------

baseline_results = pd.DataFrame({

    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ],

    "Score": [
        accuracy,
        precision,
        recall,
        f1
    ]

})

baseline_results.to_csv(
    f"{PROJECT_DIR}/reports/baseline_results.csv",
    index=False
)

display(baseline_results)


# ================================================================
# 18. CONFUSION MATRIX
# ================================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=[0, 1]
)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[
        "No Risk",
        "Low Risk"
    ],
    yticklabels=[
        "No Risk",
        "Low Risk"
    ]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.title(
    "Logistic Regression Confusion Matrix"
)

plt.tight_layout()

plt.savefig(
    f"{PROJECT_DIR}/reports/confusion_matrix.png"
)

plt.show()


# ================================================================
# 19. SAVE TRAINED MODEL
# ================================================================

joblib.dump(
    baseline_model,
    f"{PROJECT_DIR}/models/logistic_regression_baseline.pkl"
)

print("\nModel saved successfully!")


# ================================================================
# 20. SAVE CLEANING DECISIONS
# ================================================================

with open(
    f"{PROJECT_DIR}/reports/cleaning_decisions.json",
    "w"
) as file:

    json.dump(
        cleaning_log,
        file,
        indent=4
    )


# ================================================================
# 21. GENERATE  README
# ================================================================

readme = f"""
# AgriPak Farms - Low-Yield Risk Prediction

## Project Overview

Machine Learning internship project for AgriPak Farms.

The objective is to predict whether an agricultural field
is at risk of low yield before harvest, allowing timely
agricultural intervention.

## Week 1 Objectives

- Dataset loading and inspection
- Data cleaning and validation
- Reproducible stratified train-test split
- Logistic Regression baseline model
- Numerical feature correlation analysis
- Categorical feature association analysis
- Leakage sanity check
- Model evaluation and reporting

## Dataset Information

- Dataset: crop_data.csv
- Original records: {original_rows}
- Cleaned records: {cleaned_rows}
- Removed records: {removed_rows}
- Total input features: {X.shape[1]}
- Target: {TARGET_COLUMN}
- Training records: {len(X_train)}
- Testing records: {len(X_test)}
- Train-test ratio: 80/20
- Random state: {RANDOM_STATE}

## Target Classes

- 0 = No Low-Yield Risk
- 1 = Low-Yield Risk

## Data Cleaning Decisions

"""

for decision in cleaning_log:

    readme += f"- {decision}\n"

readme += """
- Missing numerical features are imputed using the training-set median.
- Missing categorical features are imputed using the training-set mode.
- Numerical features are standardized.
- Categorical features are one-hot encoded.
- Preprocessing is fitted only on training data.

## Feature Classification

### Numerical Features

"""

for feature in numeric_features:

    readme += f"- {feature}\n"

readme += """
### Categorical Features

"""

for feature in categorical_features:

    readme += f"- {feature}\n"

readme += f"""
## Correlation and Leakage Analysis

Correlation analysis was performed using training data only.

### Numerical Features

- Method: Pearson correlation with the binary target.
- Target: low_yield_risk.

### Categorical Features

- Method: One-hot encoding followed by Pearson correlation.
- Individual category correlations were calculated.
- Feature-level association is represented by the maximum
  absolute category-level correlation.

### Leakage Investigation

- Investigation threshold: Absolute correlation > {THRESHOLD}.
- No individual feature exceeding this threshold is automatically
  classified as leakage.
- High correlation alone does not establish data leakage.

"""

if not flagged_features.empty:

    readme += "### Flagged Features\n\n"

    for _, row in flagged_features.iterrows():

        feature = row["Feature"]

        score = row["Absolute_Correlation"]

        feature_type = row["Feature_Type"]

        readme += (
            f"#### {feature}\n"
            f"- Feature type: {feature_type}\n"
            f"- Maximum absolute correlation: {score:.4f}\n"
            "- Assessment: Requires manual investigation.\n\n"
        )

else:

    readme += (
        "No numerical or categorical feature exceeded "
        "the absolute correlation threshold of 0.90 "
        "in the training data.\n\n"
    )

readme += f"""
### Additional Leakage Checks

- Exact train-test overlapping records: {overlap_count}
- Train-test overlap check performed on complete records.
- Preprocessing is fitted using training data only.
- Test data is used only for final prediction and evaluation.

### Leakage Investigation Limitations

The following still require investigation:

- Original target-generation methodology.
- Feature availability at the intended prediction date.
- Potential temporal leakage.
- Potential field-level or farm-level overlap.

No correlation threshold or single overlap check proves
that the dataset is completely leakage-free.

### Generated Correlation Reports

- reports/all_feature_correlation.csv
- reports/categorical_correlation_details.csv
- reports/potential_leakage_features.csv
- reports/all_feature_correlation.png

## Baseline Model

Logistic Regression with:

- Numerical imputation using median.
- Numerical standardization.
- Categorical imputation using most frequent value.
- One-hot encoding for categorical features.
- Preprocessing fitted only on training data.

## Evaluation Results

| Metric | Score |
|---|---:|
"""

readme += f"""
| Accuracy | {accuracy:.4f} |
| Precision | {precision:.4f} |
| Recall | {recall:.4f} |
| F1 Score | {f1:.4f} |

Positive class: 1 (Low-Yield Risk).

## How to Run

1. Install dependencies:

   pip install -r requirements.txt

2. Open the project notebook in Google Colab.

3. Upload crop_data.csv.

4. Run all notebook cells.

5. Review generated model metrics and correlation reports.

## Project Structure

- data/raw: Original dataset
- data/processed: Cleaned dataset and train-test split
- src: Reusable source scripts
- notebooks: Project notebooks
- models: Trained baseline model
- reports: Evaluation and correlation reports

## Reproducibility

- Random state: {RANDOM_STATE}
- Stratified train-test split: 80/20

## Model Limitations

This is an initial baseline model.

Further validation, feature investigation, model comparison,
and evaluation on genuinely independent agricultural data
are required before production use.
"""

with open(
    f"{PROJECT_DIR}/README.md",
    "w",
    encoding="utf-8"
) as file:

    file.write(readme)

print("\nUpdated README generated successfully!")


# ================================================================
# 22. REQUIREMENTS.TXT
# ================================================================

requirements = """
pandas
numpy
scikit-learn
matplotlib
seaborn
joblib
"""

with open(
    f"{PROJECT_DIR}/requirements.txt",
    "w"
) as file:

    file.write(
        requirements.strip()
    )


# ================================================================
# 23. GITIGNORE
# ================================================================

gitignore = """
__pycache__/
*.pyc
.ipynb_checkpoints/
.env
venv/
.venv/
.DS_Store
"""

with open(
    f"{PROJECT_DIR}/.gitignore",
    "w"
) as file:

    file.write(
        gitignore.strip()
    )


# ================================================================
# 24. LICENSE
# ================================================================

license_text = """
MIT License

Copyright (c) 2026 AgriPak Farms Project Contributors

Permission is hereby granted, free of charge, to any person obtaining
a copy of this software and associated documentation files, to deal
in the Software without restriction, including without limitation
the rights to use, copy, modify, merge, publish, distribute,
sublicense, and/or sell copies of the Software.

THE SOFTWARE IS PROVIDED "AS IS", WITHOUT WARRANTY OF ANY KIND.
"""

with open(
    f"{PROJECT_DIR}/LICENSE",
    "w"
) as file:

    file.write(
        license_text.strip()
    )


# ================================================================
# 25. CREATE REUSABLE DATA CLEANING SCRIPT
# ================================================================

cleaning_script = '''
import os
import pandas as pd


def clean_dataset(input_path, output_path, target_column):

    df = pd.read_csv(input_path)

    original_rows = len(df)

    df.columns = df.columns.astype(str).str.strip()

    if target_column not in df.columns:
        raise ValueError(
            f"Target {target_column} not found."
        )

    df[target_column] = pd.to_numeric(
        df[target_column],
        errors="coerce"
    )

    missing_target = df[target_column].isna().sum()

    df = df.dropna(
        subset=[target_column]
    ).copy()

    if not set(df[target_column].unique()).issubset({0, 1}):
        raise ValueError(
            "Target must contain only 0 and 1."
        )

    df[target_column] = df[target_column].astype(int)

    if df[target_column].nunique() != 2:
        raise ValueError(
            "Both target classes must be present."
        )

    duplicates = df.duplicated().sum()

    df = df.drop_duplicates().copy()

    for col in df.columns:

        if col == target_column:
            continue

        if df[col].dtype == "object":

            converted = pd.to_numeric(
                df[col],
                errors="coerce"
            )

            count = df[col].notna().sum()

            if count > 0:

                ratio = converted.notna().sum() / count

                if ratio >= 0.90:
                    df[col] = converted

    os.makedirs(
        os.path.dirname(output_path),
        exist_ok=True
    )

    df.to_csv(
        output_path,
        index=False
    )

    print("Original records:", original_rows)
    print("Cleaned records:", len(df))
    print("Missing targets removed:", missing_target)
    print("Duplicate records removed:", duplicates)

    return df


if __name__ == "__main__":

    clean_dataset(
        "data/raw/crop_data.csv",
        "data/processed/cleaned_data.csv",
        "low_yield_risk"
    )
'''

with open(
    f"{PROJECT_DIR}/src/data_cleaning.py",
    "w"
) as file:

    file.write(cleaning_script)


# ================================================================
# 26. CREATE PROJECT SUMMARY
# ================================================================

summary = {

    "project": "AgriPak Farms - Low-Yield Risk Prediction",

    "week": 1,

    "target": TARGET_COLUMN,

    "random_state": RANDOM_STATE,

    "train_test_ratio": "80/20",

    "original_records": int(original_rows),

    "cleaned_records": int(cleaned_rows),

    "training_records": int(len(X_train)),

    "testing_records": int(len(X_test)),

    "features": X.columns.tolist(),

    "numerical_features": numeric_features,

    "categorical_features": categorical_features,

    "data_quality": {

        "missing_target_rows_removed": int(
            missing_target_count
        ),

        "duplicate_rows_removed": int(
            duplicates
        )

    },

    "leakage_investigation": {

        "correlation_data": "Training data only",

        "numerical_method": "Pearson correlation",

        "categorical_method": (
            "One-hot encoding + Pearson correlation"
        ),

        "threshold": THRESHOLD,

        "flagged_features": (
            flagged_features["Feature"].tolist()
        ),

        "exact_train_test_overlap": int(
            overlap_count
        ),

        "target_generation_verified": False,

        "temporal_leakage_verified": False,

        "field_level_overlap_verified": False

    },

    "metrics": {

        "accuracy": float(accuracy),

        "precision": float(precision),

        "recall": float(recall),

        "f1_score": float(f1)

    }

}

with open(
    f"{PROJECT_DIR}/reports/project_summary.json",
    "w"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )

print("\nProject summary generated!")


# ================================================================
# 27. FINAL PROJECT STRUCTURE
# ================================================================

print("\n========== FINAL PROJECT FILES ==========")

for root, dirs, filenames in os.walk(PROJECT_DIR):

    dirs.sort()
    filenames.sort()

    for filename in filenames:

        full_path = os.path.join(
            root,
            filename
        )

        print(
            os.path.relpath(
                full_path,
                PROJECT_DIR
            )
        )


# ================================================================
# 28. CREATE ZIP FILE
# ================================================================

zip_path = shutil.make_archive(
    "/content/AgriPak-Low-Yield-Risk-Prediction",
    "zip",
    PROJECT_DIR
)


# ================================================================
# 29. FINAL RESULTS
# ================================================================

print("\n==========================================")
print("WEEK 1 PIPELINE COMPLETED SUCCESSFULLY!")
print("==========================================")

print("\nDataset:")
print("Original records:", original_rows)
print("Cleaned records:", cleaned_rows)

print("\nLeakage Investigation:")
print("Correlation analysis: Training data only")
print("Threshold:", THRESHOLD)
print("Flagged features:", len(flagged_features))
print("Exact train-test overlap:", overlap_count)

print("\nFinal Baseline Scores:")

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")

print("\nProject ZIP:", zip_path)

# Download complete project.
files.download(zip_path)
```
